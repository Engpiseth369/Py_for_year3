Step 1. Create two lists: the days and their temperatures. Position 0 of days matches position 0 of temps.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temps = [31, 33, 29, 34, 35, 32, 30]
```

Step 2. Print the number of days with len(). Print the first day and last day. Remember: the first index is 0 and the last index is -1

```python
print("Number of days:", len(temps))
print("First day:", days[0], temps[0])
print("Last day:", days[-1], temps[-1])
```

Step 3. Print the highest, the lowest and the average temperature The average is the sum divided by the number of items.

```python
print("Highest:", max(temps))
print("Lowest:", min(temps))
average = sum(temps) / len(temps)
print("Average: ", round(average, 1))
```

Step 4. Find the day of the highest temperature. temps.index(x) gives the postion of x; use the position in days.

```python
hottest = temps.index(max(temps))
print(f"index of hottest temps: {hottest}")
print("Hottest day:", days[hottest])
```

Step 5. Print the temperatures from lowest to highest with sorted(), then printt temps again to show the original list did not change.

```python
print("Sorted :", sorted(temps))
print("Original:", temps)
print(temps.sort())
```

1. Why do we use two lists here, and why must they be in the same order?
- One list holds the days and the other holds the temperatures. Each list stores one kind of data, which keeps the code simple.
 

- They must be in the same order because they are linked by position (index). days[0] (“Mon”) goes with temps[0], days[1] with temps[1], and so on. If the order differs, a temperature gets matched to the wrong day, for example the “Hottest day” would be wrong.
2. What is the difference between sorted(temps) and temps.sort()?

```python
print(sorted(temps))
print(temps.sort())
```

- sorted(temps) unchanged the Original list but temps.sort() changed (sorted in place)
- sorted(temps) returns a new sorted list but temps.sort() return None
- sorted(temps) is used when we still need the original order but temps.sort() is used when we don’t need the original order.
