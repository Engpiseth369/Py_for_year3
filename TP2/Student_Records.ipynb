Step 1: Create a student as a tuple and print the name (index 0).


```python
student = ("Dara", 20, 3.45)
print("Name", student[0])
```

Step 2: Unpack the tuple: put its three variable in one line.

```python
name, age, gpa = student
print(name, "is", age, "years old", "GPA", gpa)
```

Step 3: Make a list of four student tuples and print each one with a for loop that unpacks the tuple.

```python
students = [("Dara", 20, 3.45), ("Sophea", 21, 3.92), ("Vibol", 19, 3.10), ("Malis", 22, 3.78)]

for name, age, gpa in students:
    print(name, age, gpa)
```

Step 4. Sort the students by GPA, bset first.key = lamda s: s[2] tells sorted() to compare the item at index 2(the GPA). The best student is the first tuple of the result.

```python
ranking = sorted(students, key=lambda s: s[2], reverse=True)
print("Best student:", ranking[0][0])
```

Step 5. Try to change the GPA inside the tuple. Python refuses; try/except catches the error so the program can continue.

```python
try:
    student[2] = 4.0
except TypeError as error:
    print("Error:", error)
```

1. Why is a tuple a good choice for one student record? When would a list be better?

            Why is a tuple a good choice for one student record?
* It is fixed. A record like ("Dara", 20, 3.45) is one unit, and the tuple stops anyone from accidentally changing it.
* Each position has a meaning. Index 0 is the name, 1 is the age, and 2 is the GPA.
* It unpacks neatly. name, age, gpa = student gives each value its own variable in one line.
       
            When would a list be better?
* When the data needs to change, such as adding, removing, or editing items.
* When the collection holds similar items that grow or shrink, like all the students.

2. What happens if you write name, age = student? Explain the error.
        Python raises:

             ValueError: too many values to unpack (expected 2)
        
      The tuple ("Dara", 20, 3.45) has three values, but only two variables are on the left. The number of variables must match the number of items exactly.

3. Why does Python print 3.1 instead of 3.10?
* Python stores 3.10 as a number, not as text, and 3.10 and 3.1 are the exact same value. Python doesn’t keep the trailing zero, so print() shows the shortest form, 3.1.

       To show two decimal places, format it as text:

```python
print(name, age, f"{gpa:.2f}")
```
