Step 1. Create the grade book: a dictionary whose values are list

```python

grades = {
    "Dara": [80, 75, 90],
    "Sophea": [95, 88, 92],
    "Vibol": [40, 55, 45],
    "Malis": [70, 85, 78],
}
```

Step 2. Compute each student's average and store it in a new dictionary averages

```python
averages = {}
for name, scores in grades.items():
    averages[name] = round(sum(scores) / len(scores), 1)
print("Averages: ", averages)
```

Step 3. The pass mark is 50. Print PASS or FAIL for each student and put the students who passed in a set.

```python
passed = set()
for name, avg in averages.items():
    if avg >= 50:
        print(name, "PASS")
        passed.add(name)
        
    else:
        print(name, "FAIL")
print("Passed:", sorted(passed))
```

Step 4. Rank the students.averages.items() gives(name, average) tuples; sort them by the average(index 1). best first, and number them with enumerate()

```python
ranking = sorted(averages.items(), key=lambda item: item[1], reverse=True)
for rank, (name, avg) in enumerate(ranking, start=1):
    print(f"#{rank} {name} ({avg})")
```

1. Name the four data structures used in this problem and say why each one was chosen.
* Dictionary (grades, averages): Each student’s name is a unique key that maps directly to their data, so you can look up “Dara” instantly without searching through a list.
* List (the three scores): The scores are an ordered collection that you want to loop over, sum, and take len() of. A list is also easy to change if a score is corrected.
* Set (passed): It holds each passing student only once, with no duplicates, and it supports fast membership checks like "Dara" in passed. Order doesn’t matter here.
* Tuple (from averages.items()): Each (name, average) pair is a fixed unit that shouldn’t be modified. Tuples let sorted() rank the pairs, and they unpack neatly in the loop with enumerate(), as in for rank, (name, avg) in ....

3. Why is passed a set and not a list?
- No duplicates: A set stores each name only once, so a student can’t be listed as passed twice by accident.
- Fast lookup: Checking name in passed is much faster with a set (it uses hashing) than with a list, which has to scan every item. This matters with large classes.
- Order doesn’t matter: The group of passers has no meaningful order, which is exactly what a set represents. That’s why the code uses sorted(passed) when it wants a tidy, alphabetical printout.
