Step 1. Create the dictionary and read the price of a mango with its key.

```python
prices = {"apple": 0.50, "banana": 0.25, "mango": 1.00}
print("Price of mange:", prices["mango"])
```

Step 2. Add oranges at $0.75, change the apple price to $0.60, and delete bananas with del.

```python
prices["orange"] = 0.75
prices["apple"] = 0.60
del prices["banana"]
print("Prices:", prices)
```

Step 3. prices["grape"] wouldd crash because the key does not exist. get() returns a default value instead.

```python
print("Price of grape:", prices.get("grape", "not sold"))
```

Step 4. Print a price list with a loop over item(), which gives each (key, value) pair.

```python
for fruit, price in prices.items():
    print(f"{fruit:<7} ${price:.2f}")
```

Step 5. A customer buys 4 apples, 2 mangoes and 3 oranges. Compute the bill: for each fruit, add price x quantity to total.

```python
basket = {"apple": 4, "mango": 2, "orange": 3}
total = 0
for fruit, qty in basket.items():
    total += prices[fruit] * qty
print(f"Total bill: ${total:.2f}")
```

1. Why is a dictionarr better that two lists (fruits and prices) for this shop?
* Direct lookup: prices["mango"] gives the price right away. With two lists you would need prices[fruits.index("mango")], which searches the whole list.
* Linked data: Each fruit is tied to its own price. With two lists, the fruit and price are only connected by matching positions.
* Safer to change: Deleting banana with del prices["banana"] removes the fruit and its price together. With two lists you must remove from both, and if you forget one, every price after it shifts to the wrong fruit.
* Easy to read: The code says what it means: key = fruit, value = price.
